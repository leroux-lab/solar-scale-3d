# 🪐 Solar Scale 3D — Interactive Python Math & Data Notebook
Since you know Python, this **Jupyter Notebook (`.ipynb`)** lets you experiment with new celestial bodies, test new **"If Earth Were a..."** object metaphors, and verify travel-time calculations before adding them to `src/planetsData.js`!

```python
# 1. Core Scale Calculator: Test any Everyday Object size for Earth!
EARTH_DIAMETER_KM = 12742.0

bodies = [
    ("Ceres (Dwarf)", 939, 413.7),
    ("Pluto (Dwarf)", 2376, 5906.4),
    ("The Moon", 3474, 149.9),
    ("Mercury", 4879, 57.9),
    ("Mars", 6779, 227.9),
    ("Venus", 12104, 108.2),
    ("Earth", 12742, 149.6),
    ("Kepler-22b (Exo)", 26758, None),
    ("Neptune", 49244, 4495.0),
    ("Uranus", 50724, 2871.0),
    ("Saturn", 116460, 1434.0),
    ("Jupiter", 139820, 778.5),
    ("WASP-17b (Exo)", 278240, None),
    ("The Sun", 1392700, 0.0),
    ("Betelgeuse (Star)", 1063000000, 0.0),
    ("UY Scuti (Star)", 2376500000, 0.0),
]

def show_scale_table(earth_object_name="Glass Marble", earth_cm=1.5):
    print(f"=== IF EARTH IS A {earth_object_name.upper()} ({earth_cm} cm) ===")
    print(f"{'Body':<20} | {'Earth Widths':<14} | {'Scaled Size':<18} | {'100-Yd Field'}")
    print("-" * 72)
    for name, diam_km, dist_mkm in bodies:
        ratio = diam_km / EARTH_DIAMETER_KM
        scaled_cm = ratio * earth_cm
        if scaled_cm >= 100000:
            size_str = f"{scaled_cm / 100000:.2f} km ({scaled_cm / 160934:.2f} mi)"
        elif scaled_cm >= 100:
            size_str = f"{scaled_cm / 100:.2f} m ({scaled_cm / 30.48:.1f} ft)"
        else:
            size_str = f"{scaled_cm:.2f} cm"
        yds = f"{(dist_mkm / 4495.0) * 100:.1f} yds" if dist_mkm is not None else "Exoplanet"
        print(f"{name:<20} | {ratio:<14.3f} | {size_str:<18} | {yds}")

show_scale_table("Glass Marble", 1.5)
```

```python
# 2. Volume Comparison Calculator: How many of Body A fit inside Body B?
def compare_worlds(small_name, small_km, large_name, large_km):
    width_ratio = large_km / small_km
    volume_ratio = width_ratio ** 3
    print(f"{large_name} vs. {small_name}:")
    print(f"  • Equator Stack: {width_ratio:,.2f} {small_name}s fit side-by-side across {large_name}")
    print(f"  • 3D Volume:     {volume_ratio:,.2f} {small_name}s fit inside {large_name}!")

compare_worlds("Earth", 12742, "Jupiter", 139820)
compare_worlds("The Sun", 1392700, "Betelgeuse", 1063000000)
compare_worlds("The Sun", 1392700, "UY Scuti", 2376500000)
```
